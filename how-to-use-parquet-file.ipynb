## 🔹 Why Parquet Instead of CSV?
#### ✅ Smaller file size (compressed storage).
#### ✅ Faster reading and writing (columnar format).
#### ✅ Better support for big data tools like Spark & Hive.

##### Let me know if you need modifications! 🚀

```python
import pandas as pd
import numpy as np

# Step 1: Create a messy dataset
data = {
    "Employee ID": [101, 102, 103, 104, 105, None],
    "Name": ["Alice", "BOB", "Charlie", None, "Eve", "Frank"], 
    "Salary": [50000, None, 70000, 80000, 55000, 72000],
    "Department": ["HR", "Engineering", "HR", "Marketing", None, "Engineering"]  
}
```

```python
data
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df.dropna(subset=["Employee ID", "Name"], inplace=True)  
df["Name"] = df["Name"].str.title()  
df["Salary"].fillna(df["Salary"].mean(), inplace=True)  
df["Department"].fillna("Unknown", inplace=True)  
```

```python
df
```

```python
df.to_parquet("data_in_parquet.parquet", engine="pyarrow", compression="snappy")
```

```python
df.isnull().sum()
```
